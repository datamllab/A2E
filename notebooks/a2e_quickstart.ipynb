## 3. (Optional) Clone full A2E repo

Only needed for production eval packages or the LLM demo. Public repo → clone works without a token. Private fork → set `CLONE_FULL_REPO = True` and optional Colab Secret **`GITHUB_TOKEN`**.


````python
CLONE_FULL_REPO = False  # set True only if you want production eval packages from A2E

import os
import shutil
import subprocess
from pathlib import Path

REPO_OWNER = "datamllab"
REPO_NAME = "A2E"
BRANCH = "main"
REPO_DIR = Path("/content/A2E")

if not CLONE_FULL_REPO:
    print("Skipped clone (CLONE_FULL_REPO=False). Demo above already ran standalone.")
else:
    def _github_token() -> str:
        token = os.environ.get("GITHUB_TOKEN", "").strip()
        if token:
            return token
        try:
            from google.colab import userdata  # type: ignore
            return userdata.get("GITHUB_TOKEN").strip()
        except Exception:
            return ""

    token = _github_token()
    public_url = f"https://github.com/{REPO_OWNER}/{REPO_NAME}.git"
    if REPO_DIR.exists():
        shutil.rmtree(REPO_DIR)

    if token:
        auth_url = f"https://x-access-token:{token}@github.com/{REPO_OWNER}/{REPO_NAME}.git"
        clone_url = auth_url
    else:
        clone_url = public_url

    print(f"git clone --depth 1 --branch {BRANCH} ...")
    rc = subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", BRANCH, clone_url, str(REPO_DIR)]
    ).returncode
    if rc != 0:
        raise RuntimeError("Clone failed — check network or set GITHUB_TOKEN for private forks.")

    head = subprocess.check_output(
        ["git", "rev-parse", "--short", "HEAD"], cwd=REPO_DIR, text=True
    ).strip()
    print(f"Cloned {REPO_OWNER}/{REPO_NAME} @ {head} (branch {BRANCH})")
````

## 5. (Optional) LLM-judged metrics

Set Colab Secret `OPENAI_API_KEY` (and optionally `OPENAI_API_BASE`, `A2E_MODEL`), then set `RUN_LLM_DEMO = True`.
Scores **conciseness** on the verbose run only (~1 LLM call).



````python
RUN_LLM_DEMO = False

if not RUN_LLM_DEMO:
    print("Skipped LLM demo (RUN_LLM_DEMO=False). CODE metrics above use production eval code.")
else:
    import json
    import sys
    from pathlib import Path

    import subprocess

    subprocess.run("pip install -q openai", shell=True, check=True)

    REPO_DIR = Path("/content/AEP")
    sys.path.insert(0, str(REPO_DIR / "eval"))

    from google.colab import userdata
    from a2e.evals.llm import LLM
    from process_values.skill_eval import make_conciseness

    api_key = userdata.get("OPENAI_API_KEY")
    model = "gpt-4o-mini"
    try:
        model = userdata.get("A2E_MODEL")
    except Exception:
        pass
    base_url = "https://api.openai.com/v1"
    try:
        base_url = userdata.get("OPENAI_API_BASE")
    except Exception:
        pass

    llm = LLM(provider="openai", model=model, api_key=api_key, base_url=base_url)
    judge = make_conciseness(llm)

    demo = json.loads((REPO_DIR / "notebooks/data/colab_demo_pair.json").read_text())
    for run in demo["runs"]:
        result = judge(
            output=run["output"],
            expected=run["expected"],
            input=run["input"],
        )
        print(run["label"], "conciseness:", result)

````
